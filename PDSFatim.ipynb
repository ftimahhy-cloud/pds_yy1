# ======================================
# numpy.linspace
# ======================================

```python
# Koding T4
# T4.1
import numpy as np
# Membuat urutan dari 0 hingga 1 dengan 5 elemen
urutan, jarak = np.linspace(0, 1, 5, retstep=True)
print("Urutan:", urutan)
print("Jarak antara elemen:", jarak)
```

```python
# T4.2 Membuat urutan bilangan bulat dari 0 hingga 10 dengan 4 elemen dan tipe data integer
urutan = np.linspace(0, 10, 4, dtype=int)
print("Urutan:", urutan)
```

```python
# T4.3 Membuat urutan dengan num=5 tanpa menggunakan endpoint dan dengan menggunakan retstep
a = np.linspace(2.0, 3.0, num=5)
b = np.linspace(2.0, 3.0, num=5, endpoint=False)
c, step = np.linspace(2.0, 3.0, num=5, retstep=True)
# Menampilkan output
print("Urutan a (dengan endpoint):")
print(a)

print("\nUrutan b (tanpa endpoint):")
print(b)

print("\nUrutan c (dengan retstep):")
print(c)


print("\nJarak antara elemen dalam urutan c:")
print(step)
```

```python
# T4.4 Visualisasi np.linspace
import numpy as np
import matplotlib.pyplot as plt
N = 8
y = np.zeros(N)
x1 = np.linspace(0, 10, N, endpoint=True)
x2 = np.linspace(0, 10, N, endpoint=False)
plt.plot(x1, y, 'o')
plt.plot(x2, y + 0.5, 'o')
plt.ylim([-0.5, 1])
(-0.5, 1)
plt.show()
```
